# STEP 2 : DATA PRE-PROCESSING #
In this step we clean and prepare the dataset by handling missing values,correcting data types,removing duplicates to ensure the data is ready for analysis

```python
# droping duplicates
df.duplicated().sum()
df = df.drop_duplicates()

```

```python
# Renaming sales columns to month names
df.rename(columns={
    'sales_month_1': 'Jan',
    'sales_month_2': 'Feb',
    'sales_month_3': 'Mar',
    'sales_month_4': 'Apr',
    'sales_month_5': 'May',
    'sales_month_6': 'Jun',
    'sales_month_7': 'Jul',
    'sales_month_8': 'Aug',
    'sales_month_9': 'Sep',
    'sales_month_10': 'Oct',
    'sales_month_11': 'Nov',
    'sales_month_12': 'Dec'
}, inplace=True)
df.columns
```

```python
# Standardize column names(Column names were already standardized.) 

df.columns
```

```python
# Creating Derived Columns
# 1. Create a list of your new month column names
month_columns=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

# 2. Calculate Total Units Sold (Across the row)
# 'axis=1' tells Pandas to sum HORIZONTALLY (Jan + Feb + Mar...)
df['Total_Sales'] = df[month_columns].sum(axis=1)

# 3. Calculate Total Revenue
# This multiplies the 'price' of the item by the 'Total_Sales' units
df['Total_Revenue'] = df['price'] * df['Total_Sales']

print("Calculation successful!")
df.head()
```

```python
# Filtering data
# Filter for products with a review score higher than 4.5
high_rated = df[df['review_score'] > 4.5]

print(f"Number of high-rated products: {len(high_rated)}")
```

```python
# Sort the high_rated list by Revenue to find the top earners
top_performers = high_rated.nlargest(5, 'Total_Revenue')

# Show only the important columns
print(top_performers[['product_name', 'category', 'review_score', 'Total_Revenue']])
```

```python
# Calculate Sales Growth (H1 vs H2) 
h1_months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
h2_months = ['Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

h1_total = df[h1_months].sum(axis=1)
h2_total = df[h2_months].sum(axis=1)

# Formula: ((New - Old) / Old) * 100
# We add a tiny 0.001 to avoid "Division by Zero" errors if H1 was 0
df['Sales_Growth_Pct'] = ((h2_total - h1_total) / (h1_total + 0.001)) * 100

# Create Price Ranges (Binning)
# This labels products based on their price
df['Price_Range'] = pd.cut(df['price'], 
bins=[0, 150, 350, float('inf')], 
labels=['Budget', 'Mid-Range', 'Premium'])

print("Advanced features created!")
df[['product_name', 'price', 'Price_Range', 'Sales_Growth_Pct']].head()
```

```python
# Group by category and calculate everything in one step
category_analysis = df.groupby('category').agg({
    'price': 'mean',          
    'Total_Sales': 'sum',     
    'Total_Revenue': 'sum'    
}).reset_index()

# Display the final summary
category_analysis
```

## EDA(Exploratory Data Analysis) & Visualisations ##
In this section, I perform Exploratory Data Analysis (EDA) to dive deep into the dataset. My goal is to move beyond looking at raw numbers and use visualizations to discover the "story" behind the sales. I have structured my analysis into three levels: Univariate, Bivariate, and Multivariate.To ensure my analysis is accurate, I also used Groupby and Pivot Tables. 

```python
# Set aesthetic style for all plots
sns.set_theme(style="whitegrid")
```

# 1. Price Distribution #
I created a histogram to see if our products are mostly budget friendly or premium.The chart shows that our product prices are spread across a wide range. Most of our items are priced in the middle, meaning we are a mid-range store rather than a luxury or discount only shop.


```python
#  UNIVARIATE ANALYSIS 
plt.figure()
sns.histplot(df['price'], kde=True, color='blue')
plt.title('Distribution of Product Prices')
plt.show()
```

# 2.Category Count #
A simple count of how many unique products exist in each category.The chart shows that our inventory is evenly spread. We aren't overreliant on just one type of product, which reduces business risk

```python
plt.figure()
sns.countplot(data=df, x='category', hue='category', palette='viridis', legend=False)
plt.title('Total Number of Products per Category')
plt.xticks(rotation=45)
plt.show()
```

# 3. Review Score Spread #
I used a box plot to find the typical customer rating and identify any low rated products that might need attention.

```python
plt.figure()
sns.boxplot(x=df['review_score'], color='teal')
plt.title('Spread of Customer Review Scores')
plt.show()
```

# 4.Revenue by Category #
I plotted this to identify which department generates the highest and most consistent revenue.This clearly shows that Books and Sports bring in the most money. Even if we sell other things, these two are our 'financial engines'.

```python
#  BIVARIATE ANALYSIS 
plt.figure()
rev_by_cat = df.groupby('category')['Total_Revenue'].sum().sort_values()
rev_by_cat.plot(kind='barh', color='teal')
plt.title('Total Revenue Generated by Category')
plt.show()
```

# 5.Price vs Total Sales #
Tests if raising the price makes people buy less.the dots are scattered everywhere. This means high prices do not scare away our customers they are willing to pay more for quality.

```python
plt.figure(figsize=(10,8))
sns.scatterplot(data=df, x='price', y='Total_Sales', alpha=0.9)
plt.title('Relationship: Product Price vs Units Sold')
plt.show()
```

# 6.Monthly Sales Trend #
I used a line chart to compare revenue across all 12 months and highlight periods of seasonal increase and decline.

```python
plt.figure()
monthly_sales = df[month_columns].sum()
plt.plot(monthly_sales.index, monthly_sales.values, marker='*', linestyle='-', color='red')
plt.xticks(rotation=45)
plt.title('Total Sales Trend Across 12 Months')
plt.show()
```

# 7.Correlation Heatmap #
I generated this to see the mathematical strength between variables like Price, Reviews, and Revenue.This shows a very strong link (0.95) between Price and Revenue. It tells us that increasing the price of a popular item is the fastest way to grow our income.

```python
# MULTIVARIATE ANALYSIS
plt.figure(figsize=(10, 8))
corr = df[['price', 'review_score', 'review_count', 'Total_Sales', 'Total_Revenue']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Key Metrics')
plt.show()
```

# 8.Review Count vs Sales by Category #
I used color coding to see if customer reviews have a bigger impact on sales in specific categories like Electronics versus Clothing.It shows that in certain categories like Electronics, having more reviews helps products sell much better

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='review_count', y='Total_Sales', hue='category')
plt.title('Impact of Reviews on Sales per Category')
plt.show()
```

# 9.Revenue Distribution by Category #
I used grouped box plots to see how consistent the revenue is for products within the same department.This shows that while Books make the most money, their earnings vary a lot. Some books sells high while others sell very little.

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='category', y='Total_Revenue')
plt.title('Revenue Variance within Categories')
plt.show()
```

# 10.Review Score vs Revenue #
I added a regression line to visually prove if higher customer satisfaction actually leads to higher earnings.The upward sloping line proves that as customer satisfaction goes up, our revenue also goes up.

```python
plt.figure(figsize=(10,6))
sns.regplot(data=df, x='review_score', y='Total_Revenue', scatter_kws={'alpha':0.3},line_kws={'color':'red','label':'Trend Line'})
plt.title('Trend Analysis: Review Score impact on Total Revenue')
plt.legend() 
plt.show()
```

After visualizing the trends, I am now generating a Final Statistical Summary. While my initial exploration gave me a basic count, this advanced aggregation allows me to see the relationship between sales volume, average price, and total revenue across all categories. I am also using a Pivot Table to specifically examine if higher customer ratings directly translate to higher revenue in different departments.

```python
# A. Groupby Summary: Proving our charts with hard numbers
category_stats = df.groupby('category').agg({
    'Total_Revenue': 'sum',
    'Total_Sales': 'sum',
    'price': 'mean'
}).sort_values(by='Total_Revenue', ascending=False)

print("Category Performance Summary")
print(category_stats)

# B. Pivot Table: Revenue cross-referenced by Review Range
df['rating_group'] = pd.cut(df['review_score'], bins=[0, 2, 4, 5], labels=['Low', 'Medium', 'High'])
revenue_pivot = df.pivot_table(index='category', columns='rating_group', values='Total_Revenue', aggfunc='sum',observed=True)

print("\n Revenue Pivot Table (Category vs Rating)")
print(revenue_pivot)
```
